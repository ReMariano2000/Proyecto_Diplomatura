# 01. Análisis Exploratorio de Datos (EDA) - Demografía y Sistema Educativo (PBA)

## 1. Definición del Problema
### 1.1 Preguntas de Investigación
El presente proyecto se enfoca en responder dos interrogantes centrales para la planificación de la Provincia de Buenos Aires:

1. **¿Cómo impactará la variación histórica de nacimientos en la matrícula escolar futura?**
2. **¿Cuántos docentes y establecimientos se requerirán bajo distintos escenarios de demanda?**

### 1.2 Formulación Analítica
- **Tipo de problema:** Regresión multivariada y análisis de series temporales.
- **Variable objetivo (Target):** Matrícula proyectada, requerimiento de cargos docentes y oferta de establecimientos.
- **Variables explicativas principales:** Cohortes históricas de nacimientos con desfases temporales (rezago biológico $t-6$ para nivel primario).

### 1.3 Taxonomía de Variables
- **Cuantitativas discretas:** Año calendario, nacidos vivos, matrícula total, cargos docentes, unidades educativas (escuelas).
- **Cuantitativas continuas:** Tasa de variación interanual de natalidad (%), ratio alumno/docente, promedio de alumnos por escuela.

```python
from pathlib import Path
import numpy as np

# 1. Definimos la carpeta donde están los CSV
ruta_nacimientos = Path("../data/raw/Nacimientos")

# Listas nativas de Python para ir guardando los resultados
lista_años = []
lista_nacimientos = []

# 2. Recorremos los años del 2010 al 2024 usando un bucle for y range
for año_corto in range(10, 25):
    año_actual = 2000 + año_corto
    nombre_archivo = f"nacweb{año_corto}.csv"
    ruta_archivo = ruta_nacimientos / nombre_archivo
    
    if ruta_archivo.exists():
        total_pba_año = 0
        
        # Abrimos el archivo como texto plano, leyendo línea por línea
        with open(ruta_archivo, mode="r", encoding="latin1") as archivo:
            # Saltamos la primera línea porque es el encabezado (PROVRES, TIPPARTO...)
            encabezado = archivo.readline()
            
            # Recorremos cada línea del archivo
            for linea in archivo:
                # Limpiamos los saltos de línea y quitamos las comillas dobles
                linea_limpia = linea.strip().replace('"', '')
                
                # Dividimos la línea por las comas para obtener los campos en una lista
                columnas = linea_limpia.split(',')
                
                # De la inspección que hicimos:
                # columnas[0] es la provincia (PROVRES)
                # columnas[-1] es el último valor, que contiene la cantidad (CUENTA)
                provincia = columnas[0].strip()
                cuenta_texto = columnas[-1].strip()
                
                # Verificamos si la fila corresponde a la Provincia de Buenos Aires (código 6 o 06)
                if provincia == "6" or provincia == "06":
                    # Sumamos al acumulador del año convirtiendo el texto a entero
                    total_pba_año += int(cuenta_texto)
        
        # Guardamos el año y el total obtenido en nuestras listas
        lista_años.append(año_actual)
        lista_nacimientos.append(total_pba_año)
    else:
        print(f"Aviso: no se encontró el archivo {nombre_archivo}")

# 3. Convertimos las listas a vectores de NumPy para poder operar estadísticamente
años_np = np.array(lista_años)
nacimientos_np = np.array(lista_nacimientos)

# 4. Mostramos los datos por pantalla
print("=== NACIMIENTOS HISTÓRICOS DE PBA (2010 - 2024) ===")
for i in range(len(años_np)):
    print(f"Año: {años_np[i]} | Nacimientos: {nacimientos_np[i]:,}")
```
